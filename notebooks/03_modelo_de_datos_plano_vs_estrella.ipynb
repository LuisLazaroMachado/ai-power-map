## 3. Comparación de opciones de modelo

Se evalúan dos opciones para estructurar los datos hacia Tableau:

| Opción | Descripción | Ventaja | Desventaja |
|---|---|---|---|
| A - Tabla plana | Un solo CSV con todas las columnas | Simple, fácil de conectar | Redundancia de datos, difícil de mantener |
| B - Esquema estrella | Tabla de hechos + 4 dimensiones | Normalizado, eficiente, estándar en Tableau y BI | Requiere joins en Tableau |

### ¿Por qué esquema estrella y no copo de nieve?
El esquema copo de nieve normaliza aún más las dimensiones 
creando subdimensiones. En nuestro caso no es necesario porque 
nuestras dimensiones son simples y Tableau trabaja mejor con 
estructuras menos complejas. El esquema estrella es el estándar 
en Business Intelligence para dashboards interactivos.

### Modelo seleccionado: Esquema Estrella (Opción B)
Se selecciona por:
- Elimina redundancia de datos categóricos
- Permite filtros eficientes en Tableau
- Es el estándar en proyectos de BI profesionales
- Las métricas confirman integridad referencial perfecta
